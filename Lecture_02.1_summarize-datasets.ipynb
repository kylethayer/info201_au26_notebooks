## Height Data

Load the height data

```r
heights_data <- read.csv("survey_data/heights-cms.csv")
```

We'll use `glimpse()` to take a glance at the dataset

```r
glimpse(heights_data)
```

Note that the values are text (like "194 cm"), so we can't take a mathematical average yet. So we'll turn it into numbers.

```r
heights_numbers_data <- parse_number(heights_data$height)
```

Now we can find the average height

```r
mean(heights_numbers_data)
```

And the median height

```r
median(heights_numbers_data)
```

And the maximum height

```r
max(heights_numbers_data)
```

And the minumum height

```r
min(heights_numbers_data)
```
